# 2 · Pathway scores and the ten pipeline runs

Every analysis in the paper is a `pathwaytheme run` over one configuration file in
`configs/`; the configurations are the record of what was run. Score matrices are
cached on the gene-set collection and the scoring backend (`results/scores/`), so
the ten runs share two ssGSEA passes, one over the 50 hallmark processes and one
over GO:BP. The first run of each collection computes the scores; the others reuse
them.

The MYCN runs read sample lists derived from Brohl *et al.*'s MYCN call, which are
written first.

```python
from pathwaytheme.api import run_pipeline

RUNS = {
    "hallmark_sampletype":      "axes of variation over the 50 hallmark processes (Fig. 1b-d)",
    "gobp_sampletype":          "the same decomposition over GO:BP",
    "gobp_diagnosis":           "each diagnosis against the other tumours (Fig. S7)",
    "gobp_tumor_normal":        "tumour, normal and cell line, pairwise (Fig. 1e, Fig. S5)",
    "gobp_matched":             "tumours against cell lines of the same diagnosis (Fig. S6)",
    "gobp_nbl_mycn":            "MYCN-amplified against non-amplified neuroblastoma tumours (Fig. S8a)",
    "hallmark_nbl_mycn":        "the same over the hallmark processes (Fig. S8b)",
    "gobp_nbl_mycn_celllines":  "the same contrast within neuroblastoma cell lines",
    "gobp_nbl_mycn_4way":       "the 2 x 2 of MYCN status and sample type (Fig. S8c)",
    "hallmark_nbl_mycn_4way":   "the same over the hallmark processes",
}
for run, what in RUNS.items():
    print(f"{run:26s} {what}")
    run_pipeline(f"configs/{run}.yaml", verbose=False)
```

The pipeline output is then gathered into the tables the paper quotes:

the top loadings per component (Table S7).
