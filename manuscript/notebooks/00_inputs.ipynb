# 0 · Inputs

The case study starts from public data and nothing else. Every file below is read
by a later notebook; none is edited by hand.

| File (under `data/raw/`) | What it is |
|---|---|
| `delivered/…tpm…tsv` | TPM over 33,967 Ensembl genes for the 915 delivered samples |
| `delivered/…pca_coords…tsv` | the sequencing run sheet for the same samples (preceded by gene-level coordinates the study does not use) |
| `gencode_v36_gene_map.tsv` | GENCODE v36 Ensembl → symbol and gene type |
| `brohl_supplement/NIHMS1759318-supplement-2.xlsx` | Brohl *et al.* (2021) Table S1: clinical annotation, library type, RNA QC and the MYCN call |
| `brohl_supplement/NIHMS1759318-supplement-3.xlsx` | Brohl *et al.* (2021) Table S2 |
| `gdc_clinical/clinical_TARGET-*.tsv` | open TARGET clinical tables from the NCI Genomic Data Commons |

GO-slim theme map) ship with the code. The inputs are deposited as one archive;

against `data/raw/MANIFEST.sha256`.
